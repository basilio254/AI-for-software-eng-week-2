## Time series modeling

### Subtask:
Choose and implement appropriate time series forecasting models. Given the panel data structure (multiple countries over time), models like ARIMA, SARIMA, or even more complex models like LSTMs could be considered. We'll need to split the data into training and testing sets based on time.


**Reasoning**:
The previous code failed because there were no features left in X after dropping columns and handling outliers. For time series analysis with panel data, we should use the pivoted dataframe `df_pivot` and treat each country's time series separately or use a model that can handle this structure. Since we are performing time series analysis, we should use time-based features and potentially country as a categorical feature. The current preprocessing steps for regression are not suitable for this time series task. I need to rethink the data preparation for time series forecasting on the `df_pivot` data. I will start by splitting the `df_pivot` data into training and testing sets based on time. I will use an earlier set of years for training and the later years for testing.



```python
# =====================================================
# PART 4: DATA PREPROCESSING FOR TIME SERIES
# =====================================================

print("\n[3] Preprocessing data for time series analysis...")

# The df_pivot dataframe has countries as index and years as columns
# Years are already strings, which is fine for plotting but might need conversion for some models.
# For time series split, we split based on the time index (columns in df_pivot)

# Define the split point (e.g., use years up to 2015 for training, 2016 and 2017 for testing)
# The columns are '1975', '1985', '1995', '2005', '2010', '2015', '2016', '2017'
train_years = ['1975', '1985', '1995', '2005', '2010', '2015']
test_years = ['2016', '2017']

# Split the data
X_train_ts = df_pivot[train_years]
X_test_ts = df_pivot[test_years]

# For time series forecasting, the 'features' are the past values, and the 'target' is the future value.
# However, for modeling each country's time series independently, we might iterate through countries
# and train a model for each.

print(f"Training set shape (Countries x Years): {X_train_ts.shape}")
print(f"Testing set shape (Countries x Years): {X_test_ts.shape}")

# Display head of the split data
print("\nTraining data head:")
display(X_train_ts.head())

print("\nTesting data head:")
display(X_test_ts.head())


# =====================================================
# PART 5: TIME SERIES MODELING
# =====================================================

print("\n[4] Implementing time series forecasting...")

# For demonstration, we will train a simple ARIMA model for a single country
# A full panel data model or iterating through all countries would be more comprehensive

from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error

# Select a country for individual modeling example
sample_country = 'United States of America'

if sample_country in df_pivot.index:
    country_ts = df_pivot.loc[sample_country].dropna() # Get the time series and drop any NaNs

    # Ensure the time series has enough data points and is not empty
    if len(country_ts) > 5: # Minimum data points to try ARIMA
        # Split the single country's time series into train and test
        # Find the index where the test years start
        split_index = country_ts.index.get_loc('2016') # Assuming '2016' is in the index

        train_data = country_ts.iloc[:split_index]
        test_data = country_ts.iloc[split_index:]

        print(f"\nModeling time series for: {sample_country}")
        print(f"Train data length: {len(train_data)}")
        print(f"Test data length: {len(test_data)}")
        print(f"Train data (first 5): {train_data.head()}")
        print(f"Test data (first 5): {test_data.head()}")


        # Fit ARIMA model (example parameters, needs proper tuning)
        # Order (p,d,q): p=lag order, d=differencing order, q=moving average order
        # These parameters often require analysis like ACF/PACF plots or auto_arima
        try:
            model = ARIMA(train_data, order=(1, 1, 0)) # Example order
            model_fit = model.fit()
            print(model_fit.summary())

            # Forecast
            # The forecast method in newer statsmodels versions takes steps
            # The index for forecasting can be specified or it forecasts next steps
            forecast_steps = len(test_data)
            forecast = model_fit.forecast(steps=forecast_steps)

            # Align forecast index with test_data index for easy comparison
            forecast.index = test_data.index

            print("\nForecast:")
            print(forecast)

            # Evaluate the model
            rmse = np.sqrt(mean_squared_error(test_data, forecast))
            print(f"\nRMSE on test data for {sample_country}: {rmse:.4f}")

            # Store results (optional, for comparison later)
            # results_ts[sample_country] = {'model': model_fit, 'forecast': forecast, 'rmse': rmse}


        except Exception as e:
            print(f"Could not fit ARIMA model for {sample_country}: {e}")
            print("Consider checking data for stationarity or trying different ARIMA orders.")

    else:
        print(f"Not enough data points for {sample_country} to fit ARIMA model.")

else:
    print(f"Sample country '{sample_country}' not found in the pivoted data.")

print("\nTime series modeling step completed.")
```

## Forecasting future emissions

### Subtask:
Use the trained model to forecast future CO2 emissions for selected countries or the overall trend.


**Reasoning**:
Select representative countries, extract their historical data, train an ARIMA model for each, and forecast future emissions, storing the results.



```python
from statsmodels.tsa.arima.model import ARIMA

# 1. Select a few representative countries for forecasting
# Selecting countries with potentially more data points based on the head of df_pivot
selected_countries_forecast = ['United States of America', 'Germany', 'Brazil']

# Specify the minimum number of data points required for forecasting
min_data_points = 8 # We have 8 years of data, so let's require all of them for now

# Specify the number of future steps to forecast
forecast_steps = 5 # Forecast 5 years into the future

all_countries_forecast_results = {}

print(f"\n[7] Forecasting future emissions for selected countries...")

for country in selected_countries_forecast:
    print(f"\n  Forecasting for {country}...")

    if country in df_pivot.index:
        # 2. Extract historical time series data
        country_ts_data = df_pivot.loc[country].dropna() # Drop any remaining NaNs

        if len(country_ts_data) < min_data_points: # Check for enough data points
            print(f"    Skipping {country}: Insufficient historical data ({len(country_ts_data)} data points).")
            continue

        # Define training data (all available historical data)
        train_data_forecast = country_ts_data

        try:
            # 3. Train an appropriate time series model (ARIMA)
            # Using a simple ARIMA(1,1,0) model for demonstration.
            # In a real application, auto_arima or ACF/PACF analysis would be used.
            model = ARIMA(train_data_forecast.astype(float), order=(1, 1, 0)) # Ensure data is float type
            model_fit = model.fit()

            # 5. Use the trained model to generate forecasts
            forecast_result = model_fit.forecast(steps=forecast_steps)

            # Store historical data and forecasts
            all_countries_forecast_results[country] = {
                'historical_data': train_data_forecast,
                'forecast': forecast_result
            }

            print(f"    ✓ Forecast generated for {country}")
            print(f"      Forecast for the next {forecast_steps} years:\n{forecast_result.round(2)}")

        except Exception as e:
            print(f"    ✗ Error forecasting for {country}: {e}")

    else:
        print(f"  Country '{country}' not found in the dataset.")

print("\nTime series forecasting for selected countries completed.")
```
